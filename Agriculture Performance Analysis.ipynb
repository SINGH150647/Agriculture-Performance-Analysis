```python
from scipy.stats import f_oneway

kharif = data[data['Season'] == 'Kharif']['Yield_Tonnes_Ha']
rabi = data[data['Season'] == 'Rabi']['Yield_Tonnes_Ha']
zaid = data[data['Season'] == 'Zaid']['Yield_Tonnes_Ha']

f_stat, p_value = f_oneway(kharif, rabi, zaid)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
data['Yield_Tonnes_Ha'].isna().sum()
```

```python
from scipy.stats import f_oneway

kharif = data[data['Season'] == 'Kharif']['Yield_Tonnes_Ha'].dropna()
rabi = data[data['Season'] == 'Rabi']['Yield_Tonnes_Ha'].dropna()
zaid = data[data['Season'] == 'Zaid']['Yield_Tonnes_Ha'].dropna()

f_stat, p_value = f_oneway(kharif, rabi, zaid)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
print("FINAL KEY FINDINGS")
print("------------------")

print("1. Kharif season has the highest average yield.")
print("2. Kharif season has the highest average production.")
print("3. Kharif has the highest average profit, while Zaid has negative average profit.")
print("4. Zaid season has the lowest water efficiency.")
print("5. Sugarcane has the highest average yield among the crops.")
print("6. Punjab has the highest average yield among the states.")
print("7. Punjab has the highest average profit among the states.")
print("8. Rainfall and yield show a very weak linear correlation.")
print("9. Fertilizer usage and yield show almost no linear correlation.")
print("10. ANOVA did not find statistically significant differences in mean yield across seasons.")
```
